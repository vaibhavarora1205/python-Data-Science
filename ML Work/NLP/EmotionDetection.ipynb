```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('../../Data Files/NLP/train.txt',sep=';',header=None,names=['text','emotion'])
df.head()
```

```python
df.isnull().sum()
```

```python
df['emotion'].unique()
```

```python
emotions=df['emotion'].unique()
emotion_numbers={}
i=0
for emo in emotions:
    emotion_numbers[emo]=i
    i=i+1

df['emotion']=df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_nums(txt):
    new=""
    for i in txt:
        if not i.isdigit():
            new=new+i
    return new

df['text']=df['text'].apply(remove_nums)
```

```python
def remove_emojis(txt):
    new=""
    for i in txt:
        if i.isascii():
            new+=i

    return new

df['text']=df['text'].apply(remove_emojis)
```
