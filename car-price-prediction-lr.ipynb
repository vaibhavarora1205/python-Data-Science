## Prediction Model

```python
X=df.drop(columns=['price'],axis=1)
y=df['price']
```

```python
X.head()
```

```python
X_one_encode=pd.get_dummies(X,columns=['model','transmission','fuelType'],drop_first=True)
```

```python
X_one_encode=X_one_encode.astype(int)
X_one_encode.head()
```

```python
from sklearn.preprocessing import LabelEncoder

columns=['model','transmission','fuelType']

Xlabel=X.copy()
label_encoders={};

for i in columns:
    le=LabelEncoder()
    Xlabel[i]=le.fit_transform(Xlabel[i].astype(str))
    label_encoders[i]=le
```

```python
Xlabel
```

```python
from sklearn.preprocessing import StandardScaler

numerical_cols=['year','mileage','tax','mpg','engineSize']
scaler=StandardScaler()
X_one_encode[numerical_cols]=scaler.fit_transform(X_one_encode[numerical_cols])
```

```python
X_one_encode
```

```python
Xlabel.columns
```

```python
Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']]=scaler.fit_transform(Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']])
```

```python
Xlabel.head()
```

```python
# model making for one hot encoding 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encode, y, test_size=0.33, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1))/(n-p-1)
print("adjusted R2 Score: ",adjusted_r2)
```

```python
# for lable encoded data

X_train, X_test, y_train, y_test = train_test_split(Xlabel, y, test_size=0.33, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```
